### Data Types

Python has several built-in data types to store different kinds of values. Understanding data types is crucial for working with data effectively.

**Integers (`int`)**

Integers are whole numbers, positive or negative, without a decimal point.

```python
# Integers (int): whole numbers
age = 30
print(f"Integer: {age}")
```

**Floating-point numbers (`float`)**

Floats are numbers with a decimal point.

```python
# Floating-point numbers (float): numbers with decimal points
price = 19.99
print(f"Float: {price}")
```

**Strings (`str`)**

Strings are sequences of characters, used for text. They are enclosed in single or double quotes.

```python
# Strings (str): sequences of characters
name = "Alice"
print(f"String: {name}")
```

**Booleans (`bool`)**

Booleans represent truth values, either `True` or `False`. They are often used in conditional statements.

```python
# Booleans (bool): True or False values
is_student = True
print(f"Boolean: {is_student}")
```

**Lists (`list`)**

Lists are ordered, mutable sequences that can hold items of different data types. They are defined using square brackets `[]`.

```python
# Lists (list): ordered, mutable sequences
fruits = ["apple", "banana", "cherry"]
print(f"List: {fruits}")
```

**Tuples (`tuple`)**

Tuples are ordered, immutable sequences. Once created, their elements cannot be changed. They are defined using parentheses `()`.

```python
# Tuples (tuple): ordered, immutable sequences
coordinates = (10, 20)
print(f"Tuple: {coordinates}")
```

**Dictionaries (`dict`)**

Dictionaries are unordered collections of key-value pairs. Each key must be unique. They are defined using curly braces `{}`.

```python
# Dictionaries (dict): unordered, mutable key-value pairs
person = {"name": "Bob", "age": 25}
print(f"Dictionary: {person}")
```

**Sets (`set`)**

Sets are unordered collections of unique elements. Duplicate elements are automatically removed. They are defined using curly braces `{}` or the `set()` constructor.

```python
# Sets (set): unordered, mutable collections of unique elements
colors = {"red", "green", "blue"}
print(f"Set: {colors}")
```

### Loops

Loops are used to repeat a block of code multiple times. They are essential for automating repetitive tasks.

**`for` loop**

The `for` loop is used to iterate over a sequence (like a list, tuple, string, or range) or other iterable objects.

```python
# for loop: iterates over a sequence (like a list, tuple, string, or range)
print("For loop:")
for i in range(5):
  print(i)
```

**`while` loop**

The `while` loop executes a block of code as long as a specified condition is true.

```python
# while loop: executes a block of code as long as a condition is true
print("\nWhile loop:")
count = 0
while count < 3:
  print(count)
  count += 1
```

### Functions

Functions are reusable blocks of code that perform a specific task. They help organize your code, make it more readable, and avoid repetition.

**Defining and Calling a Function**

You define a function using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`. You call a function by its name followed by parentheses.

```python
# Defining a function
def greet(name):
  """This function greets the person passed in as a parameter."""
  print(f"Hello, {name}!")

# Calling a function
greet("Alice")
```

**Function with a return value**

Functions can return values using the `return` keyword.

```python
# Function with a return value
def add(a, b):
  return a + b

result = add(5, 3)
print(f"Sum: {result}")
```

### Comprehensions

Comprehensions provide a concise and efficient way to create lists, dictionaries, and sets.

**List comprehension**

List comprehensions are a compact way to create lists.

```python
# List comprehension: creating a list of squares
squares = [x**2 for x in range(10)]
print(f"List comprehension (squares): {squares}")
```

**Dictionary comprehension**

Dictionary comprehensions are used to create dictionaries.

```python
# Dictionary comprehension: creating a dictionary from a list
names = ["Alice", "Bob", "Charlie"]
name_lengths = {name: len(name) for name in names}
print(f"Dictionary comprehension (name lengths): {name_lengths}")
```

**Set comprehension**

Set comprehensions are used to create sets.

```python
# Set comprehension: creating a set of even numbers
even_numbers = {x for x in range(10) if x % 2 == 0}
print(f"Set comprehension (even numbers): {even_numbers}")
```

### Classes and Objects

Classes are blueprints for creating objects. They define attributes (data) and methods (functions) that the objects created from the class will have.

```python
# Defining a class
class Dog:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def bark(self):
    print(f"{self.name} says Woof!")

# Creating an object (instance of the class)
my_dog = Dog("Buddy", 3)

# Accessing attributes and calling methods
print(f"My dog's name is {my_dog.name} and he is {my_dog.age} years old.")
my_dog.bark()
```

### Inheritance

Inheritance is a way to create a new class (a child class) based on an existing class (a parent class). The child class inherits the attributes and methods of the parent class and can also add its own or override the inherited ones.

```python
# Defining a parent class
class Animal:
  def __init__(self, species):
    self.species = species

  def eat(self):
    print("Eating...")

# Defining a child class that inherits from Animal
class Cat(Animal):
  def __init__(self, species, name):
    super().__init__(species) # Call the parent class constructor
    self.name = name

  def meow(self):
    print(f"{self.name} says Meow!")

# Creating an instance of the child class
my_cat = Cat("Felis catus", "Whiskers")

# Accessing inherited attributes and methods, and calling child class methods
print(f"My cat is a {my_cat.species}.")
my_cat.eat()
my_cat.meow()
```

### Type Hints

Type hints are annotations that indicate the expected data type of variables, function arguments, and function return values. They are not enforced at runtime but can be used by static analysis tools (like linters and type checkers) to help catch type-related errors before running the code.

### Decorators

Decorators are a powerful and flexible feature in Python that allows you to modify or enhance functions or methods. They are essentially functions that take another function as an argument and return a new function, usually extending the behavior of the original function without explicitly modifying it.

```python
# Example of a simple decorator

def my_decorator(func):
  def wrapper():
    print("Something is happening before the function is called.")
    func()
    print("Something is happening after the function is called.")
  return wrapper

@my_decorator
def say_hello():
  print("Hello!")

# Calling the decorated function
say_hello()
```

### Error Handling

Error handling is crucial for writing robust code. Python uses `try`, `except`, `else`, and `finally` blocks to manage errors (exceptions).

**`try` and `except`**

The `try` block contains the code that might raise an exception. The `except` block is executed if a specific exception occurs within the `try` block.

```python
# Example of try and except

try:
  result = 10 / 0  # This will raise a ZeroDivisionError
except ZeroDivisionError:
  print("Error: Cannot divide by zero!")
```

You can also catch multiple exceptions or a general `Exception`.

```python
# Catching multiple exceptions

try:
  num = int("abc") # This will raise a ValueError
except (ValueError, TypeError):
  print("Error: Invalid value or type!")
```

**`else` block**

The `else` block is executed if the `try` block runs without raising any exceptions.

```python
# Example of try, except, and else

try:
  result = 10 / 2
except ZeroDivisionError:
  print("Error: Cannot divide by zero!")
else:
  print(f"Result: {result}")
```

**`finally` block**

The `finally` block is always executed, regardless of whether an exception occurred or not. It's often used for cleanup operations.

```python
# Example of try, except, else, and finally

try:
  f = open("myfile.txt")
  # Perform file operations
except FileNotFoundError:
  print("Error: File not found!")
finally:
  # This will always execute, even if an error occurred
  # if 'f' is defined, close it
  if 'f' in locals() and not f.closed:
      f.close()
  print("Finishing up.")
```

### Iterators and Generators (`yield`)

**Iterators** are objects that can be iterated upon. They implement the iterator protocol, which consists of the `__iter__()` method (returns the iterator object itself) and the `__next__()` method (returns the next item in the sequence, or raises `StopIteration` when there are no more items).

**Generators** are a simple and powerful tool for creating iterators. They are written like regular functions but use the `yield` keyword instead of `return`. When a generator function is called, it returns an iterator. The `yield` statement pauses the function's execution and returns a value to the caller, and it resumes execution right after the `yield` statement when `__next__()` is called again.

```python
# Example of an iterator

my_list = [1, 2, 3, 4]
my_iterator = iter(my_list)

print("Iterating through a list:")
print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))
print(next(my_iterator))

# This would raise StopIteration
# print(next(my_iterator))
```

```python
# Example of a generator function

def count_up_to(n):
  i = 1
  while i <= n:
    yield i
    i += 1

# Using the generator
counter = count_up_to(5)

print("\nUsing a generator:")
print(next(counter))
print(next(counter))
print(next(counter))
print(next(counter))
print(next(counter))

# This would raise StopIteration
# print(next(counter))

# Generators are often used in for loops
print("\nUsing a generator in a for loop:")
for num in count_up_to(3):
  print(num)
```

### Context Managers (`__enter__`, `__exit__`)

Context managers provide a convenient way to manage resources (like files or network connections) that need setup and teardown operations. They are typically used with the `with` statement. Objects that support the context management protocol implement the `__enter__()` and `__exit__()` methods.

*   `__enter__()`: Called when the `with` statement is entered. It can return an object that is assigned to the variable after `as`.
*   `__exit__(exc_type, exc_value, traceback)`: Called when the `with` statement is exited. It handles any exceptions that occurred within the `with` block.

```python
# Example of using a built-in context manager (file handling)

print("Using a file context manager:")
try:
  with open("example.txt", "w") as f:
    f.write("Hello, world!\n")
    f.write("This is a test file.")
  # The file is automatically closed when exiting the 'with' block
  print("File 'example.txt' written successfully.")
except IOError:
  print("Error writing to file!")

# Example of creating a custom context manager

class MyContextManager:
  def __enter__(self):
    print("\nEntering the context...")
    return self # Can return an object

  def __exit__(self, exc_type, exc_value, traceback):
    print("Exiting the context...")
    if exc_type:
      print(f"An exception occurred: {exc_type}, {exc_value}")
    return False # Propagate the exception if False

# Using the custom context manager
print("Using a custom context manager:")
with MyContextManager() as mc:
  print("Inside the context...")
  # You can use the object returned by __enter__ here (if any)
  # raise ValueError("Something went wrong!") # Uncomment to see exception handling
```

**`random` module**

The `random` module implements pseudo-random number generators for various distributions.

```python
import random

# Generating a random integer within a range
random_int = random.randint(1, 10)
print(f"A random integer between 1 and 10: {random_int}")

# Generating a random floating-point number between 0 and 1
random_float = random.random()
print(f"A random float between 0 and 1: {random_float}")

# Shuffling a list
my_list = [1, 2, 3, 4, 5]
random.shuffle(my_list)
print(f"Shuffled list: {my_list}")

# Choosing a random element from a sequence
my_sequence = ["apple", "banana", "cherry"]
random_element = random.choice(my_sequence)
print(f"Random element from sequence: {random_element}")
```

### File Handling

Python provides built-in functions to work with files, allowing you to read data from files and write data to files. The `open()` function is used to open a file, and it returns a file object.

**Writing to a file**

To write to a file, open it in write mode (`"w"`). If the file doesn't exist, it will be created. If it does exist, its contents will be overwritten. Use the `write()` method to write strings to the file.

### Working with JSON Data

JSON (JavaScript Object Notation) is a lightweight data interchange format that is easy for humans to read and write and easy for machines to parse and generate. Python's built-in `json` module allows you to work with JSON data.

**Reading JSON from a file**

You can load JSON data from a file into a Python dictionary or list using the `json.load()` function. We'll use the `anscombe.json` file available in this environment as an example.

```python
import json

# Reading JSON data from a file
try:
  with open("sample_data/anscombe.json", "r") as f:
    json_data = json.load(f)

  print("\nJSON data loaded from anscombe.json (first 5 elements):")
  # Print the first few elements to avoid flooding the output
  if isinstance(json_data, list):
      for item in json_data[:5]:
          print(item)
  elif isinstance(json_data, dict):
      for key, value in list(json_data.items())[:5]:
          print(f"{key}: {value}")
  else:
      print(json_data)

except FileNotFoundError:
  print("Error: anscombe.json not found!")
except json.JSONDecodeError:
  print("Error: Could not decode JSON from the file.")
```

**Writing JSON to a file**

You can write Python dictionaries or lists to a file in JSON format using the `json.dump()` function.

```python
import json

# Writing a Python dictionary to a JSON file
python_dict = {
    "name": "Alice",
    "age": 30,
    "is_student": False,
    "courses": ["Math", "Science"]
}

with open("my_output_data.json", "w") as f:
  json.dump(python_dict, f, indent=4) # Use indent for pretty printing

print("\nPython dictionary written to my_output_data.json as JSON.")
```
