```python
colors_dict = {
    'green': 'mediumseagreen',
    'blue': 'deepskyblue',
    'red': 'firebrick',
    'white': 'snow',
    'grey': 'grey',
    'gray': 'gray',
    'black': 'black',
    'silver':'silver',
    'gold': 'gold',
    'purple': 'purple',
    'pink': 'pink',
    'yellow': 'yellow',
    'orange': 'orange',
    'orange / white': 'wheat',
    'blue-white': 'lightblue',
    'red / black': 'darkred',
}

color_counts = df["Skin color"].value_counts()
order = color_counts.index
colors = [colors_dict[color] for color in order]
```

```python
plt.figure(figsize=(10,10))
sns.countplot(x="Skin color", data=df, order=order, palette=colors, hue="Skin color", hue_order=order, legend=False)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,10))
sns.countplot(y="Skin color", data=df, order=order, palette=colors, hue="Skin color", hue_order=order, legend=False)
plt.show()
```

## Correlation Coefficients


<font size="5">
    
**<u>Definition</u>**

A correlation coefficient expresses a *statistical relationship* between two variables.

In our context, the variables are two columns of a given data set of observations.

</font>

<font size="5">
    
**<u>Values</u>**
+ values in the range [−1, +1]
+ ± 1 means perfect correlation:
    + +1 indicates a perfect uphill (positive) linear relationship
    + -1 indicates a perfect downhill (negative) linear relationship
+  0 means no correlation / linear relationship

</font>

<font size="5">
    
**<u>Types</u>**

</font>

<font size="5">
    
The **Pearson** product-moment correlation coefficient is a measure of the strength and direction of the linear relationship between two variables.

Formula:

\begin{equation*}
\rho_{X,Y} = \frac{cov(X,Y)}{\sigma_{X} \sigma_{Y}}
\end{equation*}

, where
+ X, Y are two variables
+ $cov$ is the covariance of the variables (the joint variability of the variables)
+ $\sigma$ is the standard deviation of a variable (the amount of variation of a set of values)

</font>

<font size="5">
    
*Rank correlation*

**Spearman**'s rank correlation coefficient is a measure of how well the relationship between two variables can be described by a monotonic function.

The **Kendall** tau rank correlation coefficient is a measure of the portion of ranks that match between two data sets.


</font>

### Correlation between height and weight

```python
df[["Weight", "Height"]].corr("pearson")
```

```python
df[["Weight", "Height"]].corr("spearman")
```

```python
df[["Weight", "Height"]].corr("kendall")
```

### Height and Weight distribution

```python
fig, ax = plt.subplots()
fig.set_size_inches(10, 8)
sns.scatterplot(x="Weight", y="Height", data=df, s=75)
plt.xlabel("Weight (pounds)")
plt.ylabel("Height (cm)")
```

### label the tallest heroes

```python
fig, ax = plt.subplots()
fig.set_size_inches(10, 8)
sns.scatterplot(x="Weight", y="Height", data=df, s=75, color="coral")
plt.ylim((0, 1100))
plt.xlabel("Weight (pounds)")
plt.ylabel("Height (cm)")
for _, row in df.iterrows():
    if row["Height"] > 600:
        print(row["Name"] + ": " + str(round(row["Height"] / 100, 2)) + " m")
        ax.annotate(row["Name"], (row["Weight"], row["Height"]))
```
